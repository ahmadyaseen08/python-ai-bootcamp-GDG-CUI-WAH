```python
# Day 2: NumPy & Pandas Titanic Analysis - GDG CUI Wah Bootcamp

import numpy as np
import pandas as pd

print("Libraries imported successfully!")

# NumPy Practice
marks = np.array([70, 80, 90, 60, 100])
print("Marks Array:", marks)
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))

# Pandas & Titanic Dataset Practice
# (Assuming Titanic-Dataset.csv is in the repository folder)
try:
    df = pd.read_csv("Titanic-Dataset.csv")
    print("\nDataset loaded successfully!")
    print("\nShape:", df.shape)
    print("\nMissing values:\n", df.isnull().sum())

    # Filter female passengers
    female_passengers = df.loc[df["Sex"] == "female", ["Name", "Age", "Survived"]]
    print("\nFemale Passengers Sample:\n", female_passengers.head())

    # Survival percentage
    print("\nSurvival Percentages:\n", df["Survived"].value_counts(normalize=True) * 100)
except FileNotFoundError:
    print("Titanic-Dataset.csv not found, but code structure is ready.")
```
